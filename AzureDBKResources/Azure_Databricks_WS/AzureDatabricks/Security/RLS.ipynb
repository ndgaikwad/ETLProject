```python
%sql
SELECT 
EXISTS 
(
SELECT * FROM azuredatabricks_catalog.gold.mapping_rls
WHERE email = current_user()
AND region = "west"
)
```

```python
%sql
CREATE OR REPLACE FUNCTION azuredatabricks_catalog.gold.rls_filter(p_region STRING)
RETURNS BOOLEAN 
RETURN   
EXISTS 
(
    SELECT * FROM azuredatabricks_catalog.gold.mapping_rls 
    WHERE email = current_user() 
    AND region = lower(p_region)
)
```

```python
%sql
ALTER TABLE azuredatabricks_catalog.gold.orders_rls
SET ROW FILTER azuredatabricks_catalog.gold.rls_filter ON (region)
```
