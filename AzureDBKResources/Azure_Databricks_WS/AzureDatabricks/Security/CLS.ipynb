```python
%sql
CREATE OR REPLACE FUNCTION azuredatabricks_catalog.gold.salary_mask(salary FLOAT)
RETURN CASE WHEN is_account_group_member('admins') THEN salary ELSE 0 END;
```

```python
%sql
ALTER TABLE azuredatabricks_catalog.gold.employees 
ALTER COLUMN salary
SET MASK azuredatabricks_catalog.gold.salary_mask;
```
