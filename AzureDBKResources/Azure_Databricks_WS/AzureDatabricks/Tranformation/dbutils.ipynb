```python
content = dbutils.fs.ls("abfss://raw@azuredatabricksansh.dfs.core.windows.net/standard/autoloader_raw")

for i in content:
    dbutils.fs.cp(i.path, f"abfss://raw@azuredatabricksansh.dfs.core.windows.net/autoloader_raw_new/{i.name}")
```

```python
delete_content = dbutils.fs.ls("abfss://raw@azuredatabricksansh.dfs.core.windows.net/autoloader_raw_new")
for i in delete_content:
    dbutils.fs.rm(i.path)
```
