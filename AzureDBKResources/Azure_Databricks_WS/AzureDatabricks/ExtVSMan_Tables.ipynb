## Managed Table

```python
%sql
CREATE TABLE IF NOT EXISTS azuredatabricks_catalog.dummy.managed_table
(
    id INT,
    name STRING
)
USING delta
```

```python
%sql
INSERT INTO azuredatabricks_catalog.dummy.managed_table
VALUES (1, 'John'), (2, 'Mary');
```

## External Table

```python
%sql
CREATE TABLE IF NOT EXISTS azuredatabricks_catalog.dummy.ext_table
(
    id INT,
    name STRING
)
USING delta
LOCATION 'abfss://raw@azuredatabricksansh.dfs.core.windows.net/ext_table'
```

```python
%sql
INSERT INTO azuredatabricks_catalog.dummy.ext_table
VALUES (1, 'John'), (2, 'Mary');
```
